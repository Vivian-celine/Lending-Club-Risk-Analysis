### New Columns

```python
loan['funded_ratio'] = loan['funded_amnt'] / loan['loan_amnt']
```

```python
loan['issue_d'] = pd.to_datetime(loan['issue_d'], format='%b-%y')
loan['earliest_cr_line'] = pd.to_datetime(loan['earliest_cr_line'], format='%b-%y')
loan['credit_history_age'] = (loan['issue_d'] - loan['earliest_cr_line']).dt.days / 365
```

```python
loan['payment_ratio'] = loan['installment'] / loan ['funded_amnt']
```

```python
loan[['funded_ratio', 'credit_history_age', 'payment_ratio']].head()
```

```python
loan.to_csv('loan_featured.csv', index=False)
```

```python
loan['revol_util (%)'] = loan['revol_util (%)'].astype(float)
```

```python
loan['loan_status'] = loan['loan_status'].map({'Fully Paid': 1, 'Charged Off': 0})
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in ['grade', 'sub_grade', 'home_ownership', 'verification_status', 'purpose']:
    loan[col] = le.fit_transform(loan[col])
```

```python
Leak_columns =['total_pymnt', 'total_pymnt_inv', 'total_rec_prncp', 'total_rec_int', 'last_pymnt_amnt', 'issue_d', 'earliest_cr_line']
```

```python
loan = loan.drop(Leak_columns, axis=1)
```

```python
loan.dtypes
```

```python
from sklearn.utils import resample
majority = loan[loan['loan_status'] == 1]
minority = loan[loan['loan_status'] == 0]
#Oversample minority to match majority
minority_unsampled = resample(minority,
                              replace=True,
                              n_samples=len(majority),
                              random_state=42)
#Combine back together
loan_balanced = pd.concat([majority, minority_unsampled])
print(f'Balanced dataset shape: {loan_balanced.shape}')
print(loan_balanced['loan_status'].value_counts())                            
```

```python
##### x = all columns except loan_status (the inputs)
##### y = loan_status only (the prediction)
```

```python
x = loan_balanced.drop('loan_status', axis=1)
y = loan_balanced ['loan_status']
print(f'x shape: {x.shape}')
print(f'y shape: {y.shape}')
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(f'training rows: {x_train.shape[0]}')
print(f'Testing rows: {x_test.shape[0]}')
```

### Prediction Model

#### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Build and train the model
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(x_train, y_train)

# Test the model
y_pred = lr_model.predict(x_test)

# Check accuracy
print('LOGISTIC REGRESSION')
print(f'Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}%')
print(classification_report(y_test, y_pred))
```

#### Random forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train)
y_pred_rf = rf_model.predict(x_test)

print('RANDOM FOREST')
print(f'Accuracy: {accuracy_score(y_test, y_pred_rf)*100:.2f}%')
print(classification_report(y_test, y_pred_rf))
```

#### XGBoost

```python
import xgboost
print(xgboost.__version__)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(random_state=42, eval_metric='logloss')
xgb_model.fit(x_train, y_train)
y_pred_xgb = xgb_model.predict(x_test)

print('XGBOOST')
print(f'Accuracy: {accuracy_score(y_test, y_pred_xgb)*100:.2f}%')
print(classification_report(y_test, y_pred_xgb))
```

##### Random Forest selected as final model with 97.87% accuracy.
##### Outperforming Logistic Regression (64.37%) and XGBoost (81.97%)
